```
import pandas as pd

df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.columns = df.columns.str.strip()

print(df.head())
import numpy as np

df["TotalCharges"] = df["TotalCharges"].replace(" ", np.nan)
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
df.dropna(inplace=True)
#Average monthly spend
df["AvgMonthlySpend"] = df["TotalCharges"] / (df["tenure"] + 1)
print(df[["MonthlyCharges", "AvgMonthlySpend"]].head())
#Charge difference
df["ChargeDifference"] = df["MonthlyCharges"] - df["AvgMonthlySpend"]
print(df[["ChargeDifference"]].head())
#Tenure groups
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-1 Year", "1-2 Years", "2-4 Years", "4-6 Years"]
)

print(df["TenureGroup"].value_counts())
#Total services count
services = [
    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
  "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

for col in services:
    df[col] = df[col].replace({"Yes": 1, "No": 0, "No internet service": 0, "No phone service": 0})

df["TotalServices"] = df[services].sum(axis=1)

print(df["TotalServices"].head())
#Contract risk score
contract_map = {
    "Month-to-month": 3,
    "One year": 2,
    "Two year": 1
}

df["ContractRisk"] = df["Contract"].map(contract_map)
print(df[["Contract", "ContractRisk"]].head())
print(df.columns)
df.to_csv("feature_engineered_telco.csv", index=False)
print("Feature engineered dataset saved")
```
